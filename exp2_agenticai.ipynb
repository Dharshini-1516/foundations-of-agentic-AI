```python

from transformers import pipeline

# Load Question Answering Model
qa_pipeline = pipeline(
    "question-answering",
    model="deepset/roberta-base-squad2"
)

context = """
Artificial Intelligence (AI) is transforming industries by enabling machines
to perform tasks that normally require human intelligence. AI is used in
healthcare, education, transportation, finance, and many other sectors.
Machine Learning is a subset of Artificial Intelligence.
"""

question = input("Enter your question: ")

result = qa_pipeline(
    question=question,
    context=context
)

print("\nAnswer:")
print(result["answer"])

print("\nConfidence:")
print(result["score"])
```

```python
from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)

text = """
Artificial Intelligence has rapidly become one of the most influential
technologies in the world. It enables computers to perform tasks that
traditionally required human intelligence such as image recognition,
language understanding, decision making, and problem solving.

Today AI is widely used in healthcare, banking, transportation,
cybersecurity, manufacturing, and education. Researchers are continuously
improving AI models to make them more accurate, efficient, and reliable.

Future AI systems are expected to assist humans rather than replace them,
leading to improved productivity and better decision making.
"""

summary = summarizer(
    text,
    max_length=80,
    min_length=30,
    do_sample=False
)

print("\nSummary:\n")
print(summary[0]["summary_text"])
```

```python
from transformers import pipeline

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

text = input("Enter text: ")

result = classifier(text)

print("\nSentiment:")
print(result[0]["label"])

print("Confidence:")
print(result[0]["score"])
```

```python
from transformers import pipeline
from sklearn.metrics import accuracy_score

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

texts = [
    "I love this product",
    "Worst experience ever",
    "Amazing performance",
    "I hate this app",
    "Very useful software"
]

true_labels = [
    "POSITIVE",
    "NEGATIVE",
    "POSITIVE",
    "NEGATIVE",
    "POSITIVE"
]

predictions = []

for text in texts:
    result = classifier(text)
    predictions.append(result[0]["label"])

print("Predictions:", predictions)

accuracy = accuracy_score(true_labels, predictions)

print("\nAccuracy:", accuracy)
```
